```python
from rdkit import Chem

mol = Chem.MolFromSmiles("CCO")
mol_h = Chem.AddHs(mol)
print(Chem.MolToSmiles(mol_h))
print(Chem.MolToSmiles(Chem.RemoveHs(mol_h)))
```
