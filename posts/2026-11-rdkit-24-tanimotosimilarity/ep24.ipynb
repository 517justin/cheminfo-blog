```python
from rdkit import Chem, DataStructs
from rdkit.Chem import rdFingerprintGenerator

gen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=2048)
fp = lambda s: gen.GetFingerprint(Chem.MolFromSmiles(s))
print(round(DataStructs.TanimotoSimilarity(fp("CCO"), fp("CO")), 3))
```

```python
# 換你試：咖啡因和可可鹼、咖啡因和茶鹼
print(round(DataStructs.TanimotoSimilarity(fp("CN1C=NC2=C1C(=O)N(C(=O)N2C)C"), fp("Cn1cnc2c1c(=O)[nH]c(=O)n2C")), 3))
print(round(DataStructs.TanimotoSimilarity(fp("CN1C=NC2=C1C(=O)N(C(=O)N2C)C"), fp("Cn1c(=O)c2[nH]cnc2n(C)c1=O")), 3))
```
