```python
from rdkit import Chem

oh = Chem.MolFromSmarts("[OX2H]")
glucose = Chem.MolFromSmiles("OC[C@H]1OC(O)[C@H](O)[C@@H](O)[C@@H]1O")
matches = glucose.GetSubstructMatches(oh)
print(matches)
print(len(matches))
```

```python
# 換你試：蔗糖（砂糖）和甘油
for smi in ["OC[C@H]1O[C@@](CO)(O[C@H]2O[C@H](CO)[C@@H](O)[C@H](O)[C@H]2O)[C@@H](O)[C@@H]1O", "OCC(O)CO"]:
    print(len(Chem.MolFromSmiles(smi).GetSubstructMatches(oh)))
```
