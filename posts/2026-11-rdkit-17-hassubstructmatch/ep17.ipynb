```python
from rdkit import Chem

oh = Chem.MolFromSmarts("[OX2H]")   # 代表 OH 的搜尋條件
for smi in ["CCO", "CN1C=NC2=C1C(=O)N(C(=O)N2C)C"]:
    print(smi, Chem.MolFromSmiles(smi).HasSubstructMatch(oh))
```

```python
# 換你試：甲醇
Chem.MolFromSmiles("CO").HasSubstructMatch(oh)
```
