```python
from rdkit import Chem

for smi in ["OC[C@H]1OC(O)[C@H](O)[C@@H](O)[C@@H]1O", "CN1C=NC2=C1C(=O)N(C(=O)N2C)C"]:
    mol = Chem.MolFromSmiles(smi)
    print(mol.GetRingInfo().NumRings())
```

```python
# 換你試：蔗糖（砂糖）有幾個環？
sucrose = Chem.MolFromSmiles("OC[C@H]1O[C@@](CO)(O[C@H]2O[C@H](CO)[C@@H](O)[C@H](O)[C@H]2O)[C@@H](O)[C@@H]1O")
print(sucrose.GetRingInfo().NumRings())
```
