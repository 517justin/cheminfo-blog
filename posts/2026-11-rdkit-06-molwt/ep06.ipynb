```python
from rdkit import Chem
from rdkit.Chem import Descriptors, rdMolDescriptors

mol = Chem.MolFromSmiles("CCO")
print(rdMolDescriptors.CalcMolFormula(mol))
print(round(Descriptors.MolWt(mol), 2))
```

```python
# 換你試：葡萄糖、果糖、蔗糖的分子量
for smi in ["OC[C@H]1OC(O)[C@H](O)[C@@H](O)[C@@H]1O", "OCC(=O)[C@@H](O)[C@H](O)[C@H](O)CO", "OC[C@H]1O[C@@](CO)(O[C@H]2O[C@H](CO)[C@@H](O)[C@H](O)[C@H]2O)[C@@H](O)[C@@H]1O"]:
    print(round(Descriptors.MolWt(Chem.MolFromSmiles(smi)), 2))
```
