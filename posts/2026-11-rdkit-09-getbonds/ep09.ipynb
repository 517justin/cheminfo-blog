```python
from rdkit import Chem

for smi in ["CCO", "O=C=O", "C#C"]:
    mol = Chem.MolFromSmiles(smi)
    print(smi, [str(b.GetBondType()) for b in mol.GetBonds()])
```

```python
# 換你試：苯和丁二烯
for smi in ["c1ccccc1", "C=CC=C"]:
    mol = Chem.MolFromSmiles(smi)
    print(smi, [str(b.GetBondType()) for b in mol.GetBonds()])
```
