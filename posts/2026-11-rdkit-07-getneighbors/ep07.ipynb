```python
from rdkit import Chem

mol = Chem.MolFromSmiles("CCO")
atom = mol.GetAtomWithIdx(1)
print(atom.GetSymbol(), atom.GetDegree())
print([n.GetIdx() for n in atom.GetNeighbors()])
print(atom.GetTotalNumHs())
```

```python
# 換你試：醋酸的 1 號碳，雙鍵相連的氧只算一個鄰居
acetic = Chem.MolFromSmiles("CC(=O)O")
print([n.GetIdx() for n in acetic.GetAtomWithIdx(1).GetNeighbors()])
```
